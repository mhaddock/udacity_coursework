## Drawing Conclusions

```python
import pandas as pd
import numpy as np

jud_data = pd.read_csv('judicial_dataset_predictions.csv')
par_data = pd.read_csv('parachute_dataset.csv')
```

```python
jud_data.head()
```

```python
par_data.head()
```

```python
jud_actual_innocent = jud_data[jud_data['actual'] == 'innocent']
jud_actual_guilty = jud_data[jud_data['actual'] == 'guilty']
jud_type1Errors = jud_actual_innocent[jud_actual_innocent['predicted'] == 'guilty']
jud_type2Errors = jud_actual_guilty[jud_actual_guilty['predicted'] == 'innocent']

print(f'judgements: {jud_data.shape[0]}')
print(f'errors: {(jud_type1Errors.shape[0] + jud_type2Errors.shape[0]) / jud_data.shape[0]:.2%}')
print(f'type 1 errors: {jud_type1Errors.shape[0]}: {jud_type1Errors.shape[0] / jud_data.shape[0]:.2%}')
print(f'type 2 errors: {jud_type2Errors.shape[0]}: {jud_type2Errors.shape[0] / jud_data.shape[0]:.2%}')

print(f'type 1 errors if all predicted guilty: {jud_actual_innocent.shape[0] / jud_data.shape[0]:.2%}')
```

```python
par_actual_opens = par_data[par_data['actual'] == 'opens']
par_actual_fails = par_data[par_data['actual'] == 'fails']
par_type1Errors = par_actual_fails[par_actual_fails['predicted'] == 'opens']
par_type2Errors = par_actual_opens[par_actual_opens['predicted'] == 'fails']

print(f'jumps: {par_data.shape[0]}')
print(f'errors: {(par_type1Errors.shape[0] + par_type2Errors.shape[0]) / par_data.shape[0]:.5f}')
print(f'type 1 errors: {par_type1Errors.shape[0]}: {par_type1Errors.shape[0] / par_data.shape[0]:.5f}')
print(f'type 2 errors: {par_type2Errors.shape[0]}: {par_type2Errors.shape[0] / par_data.shape[0]:5f}')

print(f'type 2 errors if all predicted fail: {par_actual_opens.shape[0] / par_data.shape[0]:5f}')
```
