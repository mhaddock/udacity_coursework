### Homepage Experiement Data

```python
import pandas as pd
import numpy as np
import dateutil as du

df = pd.read_csv('homepage_actions.csv')
df.head()
```

```python
numActions = df.shape[0]
print(f'number of actions: {numActions}')

uniqueUsers = df.drop_duplicates(['id'], keep='last')
print(f'num of unique users: {uniqueUsers.shape[0]}')

controlUsers = uniqueUsers.query("group == 'control'")
experimentUsers = uniqueUsers.query("group == 'experiment'")

print(f'num users in control: {controlUsers.shape[0]}')
print(f'num users in experiment: {experimentUsers.shape[0]}')
```

```python
timestamps = df['timestamp'].convert_dtypes()
print(timestamps.describe())
firstInteractionStr = timestamps[timestamps.idxmin()]
lastInteractionStr = timestamps[timestamps.idxmax()]

print(f'first interaction: {firstInteractionStr}')
print(f'last interaction: {lastInteractionStr}')

firstInteraction = du.parser.parse(firstInteractionStr)
lastInteraction = du.parser.parse(lastInteractionStr)

print(lastInteraction - firstInteraction)
```
