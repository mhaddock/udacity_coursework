## Overall Shooting Statistics
On this slide, calculate and report the:

- Overall probability of a make (result = TRUE)
    - **49.02%**
- Overall probability of a miss (result = FALSE)
    - **50.98%**
- Overall proportion of three-pointers (shot_type = 3)
    - **54.81%**
- Overall proportion of two-pointers (shot_type = 2)
    - **45.19%**

**Calculations**
- $P(make) = \frac{\normalsize numMade}{\normalsize numShots}$
- $P(\neg{make}) = 1 - P(make)$
- $P(threePointer) = \frac{\normalsize numThreePointers}{\normalsize numShots}$
- $P(twoPointer) = \frac{\normalsize numTwoPointers}{\normalsize numShots}$

```python
import numpy as np
import pandas as pd
from scipy import stats

shot_data = pd.read_csv('./data/stephen_curry_shots_2023.csv')

nShots = shot_data.index.size
nMake = shot_data[shot_data.result == True].index.size
pMake = nMake / nShots

threePointers = shot_data[shot_data.shot_type == 3]
twoPointers = shot_data[shot_data.shot_type == 2]
nThreePointer = threePointers.index.size
pThreePointer = nThreePointer / nShots
nTwoPointer = twoPointers.index.size
pTwoPointer = nTwoPointer / nShots

print(f'P(make) = {pMake}, {pMake:.2%}')
print(f'P(notMake) = {1 - pMake}, {1 - pMake:.2%}')
print(f'P(threePointer) = {pThreePointer}, {pThreePointer:.2%}')
print(f'P(twoPointer) = {pTwoPointer}, {pTwoPointer:.2%}')
```

## Conditional Probabilities - Future
On this slide, use previously calculated probabilities to calculate and report the following conditional probabilities, which will require Bayes Theorem:

- If the next shot Steph shoots is a three-pointer...
    - What is the probability he makes it?
        - **41.86%**
    - What is the probability it was taken while his team had the lead (lead = True)?
        - **49.36%**
- If the next shot Steph shoots is a two-pointer...
    - What is the probability he makes it?
        - **57.72%**
    - What is the probability it was taken while his team had the lead (lead = True)?
        - **51.54%**

**Calculations:** _Bayes theorem was requested, but a simpler calculation is possible. Both are included below_

- If the next shot Steph shoots is a three-pointer...
    - $P(make \mid threePointer)$
        - **Bayes:** $\frac{\normalsize P(threePointer \mid make) \times P(make)}{\normalsize P(threePointer)}$
       - **Simple:** $\frac{\normalsize numMadeThreePointer}{\normalsize numThreePointer}$
    - $P(lead \mid threePointer)$
        - **Bayes:** $\frac{\normalsize P(threePointer \mid lead) \times P(lead)}{\normalsize P(threePointer)}$
        - **Simple:** $\frac{\normalsize numThreePointerWhileLead}{\normalsize numThreePointer}$
- If the next shot Steph shoots is a two-pointer...
    - $P(make \mid twoPointer)$
        - **Bayes:** $\frac{\normalsize P(twoPointer \mid make) \times P(make)}{\normalsize P(twoPointer)}$
       - **Simple:** $\frac{\normalsize numMadeTwoPointer}{\normalsize numTwoPointer}$
    - $P(lead \mid twoPointer)$
        - **Bayes:** $\frac{\normalsize P(twoPointer \mid lead) \times P(lead)}{\normalsize P(twoPointer)}$
        - **Simple:** $\frac{\normalsize numTwoPointerWhileLead}{\normalsize numTwoPointer}$


```python
print('\n3-pointer\nwhat is the probability he makes it?')
print('P(makesShot | threePointer)')
nMade_threePointer = threePointers[threePointers.result == True].index.size
pThreePointerWhenMadeSimple = nMade_threePointer / nMake

# Bayes
pMadeIfThreePointerBayes = (pThreePointerWhenMadeSimple * pMake)/pThreePointer
print(
    f'Bayes: {pMadeIfThreePointerBayes}, {pMadeIfThreePointerBayes:.2%}')

# Simple
pMadeIfThreePointerSimple = nMade_threePointer / nThreePointer
print(
    f'Simple: {pMadeIfThreePointerSimple}, {pMadeIfThreePointerSimple:.2%}')

print('\nwhat is the probability it was taken while his team had the lead')
print('P(lead | threePointer)')
leadShots = shot_data[shot_data.lead == True]
nLead = leadShots.index.size
pLead = nLead / nShots

nThreePointerWhileLead = leadShots[leadShots.shot_type == 3].index.size
pThreePointerWhenLeadSimple = nThreePointerWhileLead / nLead

# Bayes
pLeadWhenThreePointerBayes = (pThreePointerWhenLeadSimple * pLead) / pThreePointer
print(f'Bayes: {pLeadWhenThreePointerBayes}, {pLeadWhenThreePointerBayes:.2%}')

# Simple
pLeadWhenThreePointerSimple = nThreePointerWhileLead / nThreePointer
print(
    f'Simple: {pLeadWhenThreePointerSimple}, {pLeadWhenThreePointerSimple:.2%}')

print('\n2-pointer\nwhat is the probability he makes it?')
print('P(makesShot | twoPointer)')
nMade_twoPointer = twoPointers[twoPointers.result == True].index.size
pTwoPointerWhenMadeSimple = nMade_twoPointer / nMake

# Bayes
pMadeIfTwoPointerBayes = (pTwoPointerWhenMadeSimple * pMake) / pTwoPointer
print(
    f'Bayes: {pMadeIfTwoPointerBayes}, {pMadeIfTwoPointerBayes:.2%}')

# Simple
pMadeIfTwoPointerSimple = nMade_twoPointer / nTwoPointer
print(
    f'Simple: {pMadeIfTwoPointerSimple}, {pMadeIfTwoPointerSimple:.2%}')

print('\nwhat is the probability it was taken while his team had the lead')
print('P(twoPointer | lead)')
nTwoPointerWhileLead = leadShots[leadShots.shot_type == 2].index.size
pTwoPointerWhenLeadSimple = nTwoPointerWhileLead / nLead

# Bayes
pLeadWhenTwoPointerBayes = (pTwoPointerWhenLeadSimple * pLead) / pTwoPointer
print(f'Bayes: {pLeadWhenTwoPointerBayes}, {pLeadWhenTwoPointerBayes:.2%}')

# Simple
pLeadWhenTwoPointerSimple = nTwoPointerWhileLead / nTwoPointer
print(f'Simple: {pLeadWhenTwoPointerSimple}, {pLeadWhenTwoPointerSimple:.2%}')
```

**Calculations**
It was necessary to calculate the answer: $P(threePointer \mid make)$ in order to calculate the Bayesian solution to the previous question. This seems problematically circular, but the math checks out when verified against the simpler calculation

- If Steph just make a shot, what is the probability it was a three-pointer?
    - $P(threePointer \mid make)$
        - **Bayes:** $\frac{\normalsize P(make \mid threePointer) \times P(threePointer)}{\normalsize P(make)}$
        - **Simple:** $\frac{\normalsize numMadeThreePointers}{\normalsize numMade}$
- What is the probability it was a two-pointer?
    - $P(twoPointer \mid make)$
        - **Bayes:** $\frac{\normalsize P(make \mid twoPointer) \times P(twoPointer)}{\normalsize P(make)}$
        - **Simple:** $\frac{\normalsize numMadeTwoPointers}{\normalsize numMade}$


```python
print('P(threePointer | make)')
pThreePointerWhenMadeBayes = (pMadeIfThreePointerBayes * pThreePointer) / pMake
pThreePointerWhenMadeSimple = nMade_threePointer / nMake
print(f'Bayes: {pThreePointerWhenMadeBayes}, {pThreePointerWhenMadeBayes:.2%}')
print(f'Simple: {pThreePointerWhenMadeSimple}, {pThreePointerWhenMadeSimple:.2%}')

print('\nP(twoPointer | make)')
pTwoPointerWhenMadeBayes = (pMadeIfTwoPointerBayes * pTwoPointer) / pMake
pTwoPointerWhenMadeSimple = nMade_twoPointer / nMake
print(f'Bayes: {pTwoPointerWhenMadeBayes}, {pTwoPointerWhenMadeBayes:.2%}')
print(f'Simple: {pTwoPointerWhenMadeSimple}, {pTwoPointerWhenMadeSimple:.2%}')
```
